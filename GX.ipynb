# I. Import Libraries

```python
import great_expectations as gx
import pandas as pd
```

# II. Instantiate Data Context

```python
from great_expectations.data_context import FileDataContext

context = FileDataContext.create(project_root_dir='./')
```

# III. Connect to a Datasource

```python
context.sources.delete("csvm3data")
```

```python
# Give a name to a Datasource. This name must be unique between Datasources.
datasource_name = 'csvm3data'
datasource = context.sources.add_pandas(datasource_name)

# Give a name to a data asset
asset_name = 'trip-january'
path_to_data = 'P2M3_Stefano_Veronigo_Wijaya_data_clean.csv'
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)

# Build batch request
batch_request = asset.build_batch_request()
```

# IV. Create Expectation Suite

```python
# Creat an expectation suite
expectation_suite_name = 'm3expectation'
context.add_or_update_expectation_suite(expectation_suite_name)

# Create a validator using above expectation suite
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

# Check the validator
validator.head()
```

# V. Expectations

## V.I. Expect Column Values To Be Unique

```python
# Column `row_id` must be unique

validator.expect_column_values_to_be_unique('row_id')
```

## V.II. Expect Column Values To Be Between

```python
# Column `sales` must be between 0 and 100000

validator.expect_column_values_to_be_between(
    'sales',
    min_value=0,
    max_value=100000
)
```

## V.III. Expect Column Values To Be In Set 

```python
# Column `segment` must contain one of the following: ['Consumer', 'Corporate', 'Home Office']

validator.expect_column_values_to_be_in_set(
    'segment',
    ['Consumer', 'Corporate', 'Home Office']
)
```

## V.IV. Expect Column Values To Be In Type List

```python
# Column `profit` must be in form of integer or float

validator.expect_column_values_to_be_in_type_list(
    'profit',
    ['int64', 'float64']
)
```

## V.V. Expect Column Most Common Value To Be In Set

```python
# Column `segment` has 'Consumer' as the most frequent value

validator.expect_column_most_common_value_to_be_in_set(
    'segment',
    ['Consumer']
)
```

## V.VI. Expect Column Value Lengths To Be Between

```python
# Column `customer_name` has an expected value length of 3-50

validator.expect_column_value_lengths_to_be_between(
    'customer_name',
    min_value=3,
    max_value=50
)
```

## V.VII. Expect Table Row Count To Be Between

```python
# Row count between 1-9999

validator.expect_table_row_count_to_be_between(
    min_value=1,
    max_value=9999
)
```

# VI. Save into Expectation Suite

```python
# Save into Expectation Suite

validator.save_expectation_suite(discard_failed_expectations=False)
```

# VII. Checkpoint

Provide convenient abstraction for bundling the Validation of data batch against Expectation Suite for further actions.

```python
# Create checkpoint

checkpoint_1 = context.add_or_update_checkpoint(
    name = 'checkpoint_1',
    validator = validator,
)
```

```python
# Run checkpoint

checkpoint_result = checkpoint_1.run()
```
